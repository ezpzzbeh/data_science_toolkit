**5. Create two random numpy array of size n x p and p x m and then compute the dot product**

```python
n = 8
p = 2
m = 5

array_A = np.random.rand(n,p)
array_B = np.random.rand(p,m)

#1st way
dot_product = np.dot(array_A, array_B)
print("array_A")
print(array_A)
print("array_B")
print(array_B)

print("Dot Product: ")
print(dot_product)

#2nd way
print("Dot Product 2: ")
print(array_A @ array_B)

print("\nResult Shape:", dot_product.shape) #tell how many row and column
```
